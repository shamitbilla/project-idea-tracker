```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
text = "I love google yayy"
```

```python
text
```

```python
embeddings = model.encode(text)
embeddings
```

```python
embeddings.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    "I love Pizza",
    "Pizza is my favorite field.",
    "I enjoy Italy."
]

embeddings = model.encode(sentences)
similarity_matrix = cosine_similarity(embeddings)

print(f"Similarity(0,1): {similarity_matrix[0][1]:.4f}")
print(f"Similarity(1,2): {similarity_matrix[1][2]:.4f}")
print(f"Similarity(0,2): {similarity_matrix[0][2]:.4f}")
```
